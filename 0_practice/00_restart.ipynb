```python
# Check whether a number is even or odd.

num = 213
if num % 2 == 0:
    print("even")
elif num % 2 != 0:
    print("odd")
else: 
    print("please write a number")
```

```python
# Find the largest of three numbers.

a = [23, 56, 30]
b = sorted(a)
print(b[-1])  # Prints 56
print(max(a)) # both works
```

```python
year = 1000
# Century years must be divisible by 400
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("leap year")
else:
    print("normal year")
```

```python
# Reverse an integer (e.g., 12345 → 54321).
num = 12345
reversed_num = int(str(num)[::-1])
print(reversed_num)  # Output: 54321
```
